```python
class RandomUserItemMatrix:
    def __init__(self, num_users = 50,num_items=20, density = 0.05):
        '''
        sparse matrix dimensions

        '''
        self.num_users = num_users
        self.num_items = num_items
        self.density = density
        self.num_ratings = int(self.num_users * self.num_items * self.density)
        self.coo_matrix: scipy.sparse | None = None


    def generateMatrix(self):
        # Generate random user indices and item indices
        user_indices = np.random.randint(0, self.num_users, size=self.num_ratings)
        item_indices = np.random.randint(0, self.num_items, size=self.num_ratings)

        # Generate random ratings between 1 and 5
        ratings = np.random.randint(1, 6, size=self.num_ratings)

        # Create sparse matrix in COO format
        self.coo_matrix = coo_matrix((ratings, (user_indices, item_indices)), shape=(self.num_users, self.num_items))

        # Convert to CSR format
        user_item_matrix_csr = self.coo_matrix.tocsr()

        return user_item_matrix_csr
    
    
    def getMatinPandasFormat(self):
        # Convert to a sparse pandas DataFrame
        return pd.DataFrame.sparse.from_spmatrix(self.coo_matrix)
```
